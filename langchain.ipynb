```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

api_key = os.getenv("MAAS_API_TOKEN")
if not api_key:
    raise ValueError("Please set the MAAS_API_TOKEN environment variable.")

# 1. Initialize ChatOpenAI pointing to your custom MaaS endpoint
llm = ChatOpenAI(
    model="gpt-oss-20b",  # Replace with your endpoint's target model name
    api_key=api_key,
    base_url="https://maas.apps.rosa.test.uhxi.p3.openshiftapps.com/v1",
    temperature=0.7,
)

# 2. Define a prompt template
prompt = ChatPromptTemplate.from_template("Tell me a short joke about {topic}.")

# 3. Combine them into an LCEL chain
chain = prompt | llm

# 4. Run the chain
response = chain.invoke({"topic": "Python programming"})

# 5. Output the result
print(response.content)
```
